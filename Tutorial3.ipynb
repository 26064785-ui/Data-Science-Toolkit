6 and 8. Which columns and rows have missing (null) values?

```python
missing = df.isnull().sum()
print("Missing values in each column:")
print("missing")
print("\nColumns with any missing values:")
print(missing[missing > 0])
print("\nNumber of columns with missing values:", (missing > 0).sum())
```

```python
rows_with_missing_values = df.isnull().any(axis=1)
print("\nRows with missing values:")
print(df[rows_with_missing_values])
print("\nNumber of rows with missing values:", rows_with_missing_values.sum())
```
