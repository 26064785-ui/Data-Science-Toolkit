```python
# 1. Create a numpy array of size 2 x 2

import numpy as np
array_2x2 = np.array ([
     [1,2],
     [3,4]
 ])
print("this is my numpy 2x2 array")
print(array_2x2)
```

```python
# 2. Create a random numpy array of size 3 x 4

random3x4 = np.random.rand(3,4)

print("this is my random 3x4 array")
print(random3x4)
```

```python
# 3. Create a numpy array of ones and a numpy array of zeros

ones_array = np.ones ((3,3))
print("this is my ones array")
print(ones_array)

zeros_array = np.zeros ((3,3))
print("this is my zeros array")
print(zeros_array)
```

```python
# 4. Create a random numpy array of size 6x3 and transpose it
random6x3 = np.random.rand(6,3)
random3x6 = random6x3.T
print("this is my random 6x3 array")
print(random6x3)
```

```python
# 5. Create two random numpy array of size n x p and p x m and then compute the dot product
randomnxp = np.random.rand(3,4)
randompxm = np.random.rand(4,2)

dot_product = np.dot(randomnxp, randompxm)
print("this is my dot product")
print(dot_product)
```
